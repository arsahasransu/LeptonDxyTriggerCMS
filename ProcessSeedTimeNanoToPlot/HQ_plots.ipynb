```python
from datetime import datetime
import numpy as np

from IPython.display import IFrame, Image, display
import ROOT

ROOT.gROOT.SetBatch(True)
```

```python
datafilename = './data_histos_newseedtimeplot.root'
datafile = ROOT.TFile(datafilename)

filenameslist = ['./hists/hist_3cm.root', './hists/hist_30cm.root', './hists/hist_3m.root']
open_infiles = [ROOT.TFile.Open(filename) for filename in filenameslist]
```

```python
legendname = ['3 cm', '30 cm', '3 m']
colors = [ROOT.kRed-8, ROOT.kRed-6, ROOT.kRed]
markerstyle = [25, 26, 46]

basetextsize = 0.05
```

```python
# Finalise the EB plot
datahist = [datafile.Get('eleEBZ_el0_seedtime')+datafile.Get('eleEBZ_el1_seedtime')]

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")
binedges = np.array(np.arange(-3, 5.2, 0.2), dtype=np.float64)

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 800, 700)
ROOT.gStyle.SetOptStat(0)
ROOT.gStyle.SetLineWidth(2)
c.SetLeftMargin(0.17)
c.SetRightMargin(0.025)
c.SetBottomMargin(0.14)
c.SetTopMargin(0.075)
c.SetLogy()

legend = ROOT.TLegend(0.66, 0.65, 0.89, 0.9)
legend.SetBorderSize(0)
legend.SetTextSize(basetextsize)
normalise = True

var = 'genmatched_ph_seedtime_eb'
varhists = [file.Get(var) for file in open_infiles]

for i, hist in enumerate(datahist):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetTitle('')
    newhist.SetLineColor(ROOT.kBlack)
    newhist.SetLineWidth(2)
    newhist.SetMarkerStyle(20)
    newhist.SetMarkerSize(1.05)
    newhist.SetMarkerColor(ROOT.kBlack)

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.SetMinimum(2e-4)
    newhist.SetMaximum(1.1)

    newhist.GetXaxis().SetTitle("e/#gamma seed time [ns]")
    newhist.GetXaxis().SetTitleSize(1.3*basetextsize)
    newhist.GetXaxis().SetTitleOffset(1.0)
    newhist.GetXaxis().SetLabelSize(1.3*basetextsize)
    newhist.GetXaxis().SetLabelOffset(0.005)
    newhist.GetYaxis().SetTitle("Events / 0.2 ns")
    newhist.GetYaxis().SetTitleSize(1.3*basetextsize)
    newhist.GetYaxis().SetTitleOffset(1.3)
    newhist.GetYaxis().SetLabelSize(1.3*basetextsize)
    newhist.GetYaxis().SetLabelOffset(0.005)

    newhist.Draw("e1")

    legend.AddEntry(newhist, "Z #rightarrow ee, #bf{2023}", "lep")

for i, hist in enumerate(varhists):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetLineColor(colors[i])
    newhist.SetLineWidth(3)
    newhist.SetMarkerStyle(markerstyle[i])
    newhist.SetMarkerSize(1.0)
    newhist.SetMarkerColor(colors[i])

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.Draw("e1 same")

    legend.AddEntry(newhist, f'c#tau = {legendname[i]}', "lpe")

text = ROOT.TLatex()
text.SetTextSize(1.3*basetextsize)
text.SetTextFont(42)
text.SetNDC()
text.DrawLatex(0.21, 0.845, "#bf{CMS}")
text.SetTextSize(basetextsize)
text.DrawLatex(0.21, 0.78, "#chi^{#pm} #rightarrow #chi^{0} l^{#pm} #nu")
text.SetTextSize(basetextsize)
text.DrawLatex(0.21, 0.715, "|#eta| < 1.4")
text.SetTextSize(basetextsize)
text.SetTextAlign(31)
text.DrawLatex(0.975, 0.945, "0.41 fb^{-1} (13.6 TeV)")

legend.Draw()

line1 = ROOT.TLine(-3, 1.1, 5, 1.1)
line1.SetLineColor(ROOT.kBlack)
line1.SetLineWidth(2)
line1.Draw("SAME")

line2 = ROOT.TLine(5, 2e-4, 5, 1.1)
line2.SetLineColor(ROOT.kBlack)
line2.SetLineWidth(2)
line2.Draw("SAME")

c.Update()

plotname = f"plots/tdelay_ecal_w_d0_eb_new.pdf"
c.SaveAs(plotname)
```

```python
# Finalise the EE plot

datahist = [datafile.Get('eleEEZ_el0_seedtime')+datafile.Get('eleEEZ_el1_seedtime')]

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")
binedges = np.array(np.arange(-2.4, 3.6, 0.2), dtype=np.float64)

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 800, 700)
ROOT.gStyle.SetOptStat(0)
ROOT.gStyle.SetLineWidth(2)
c.SetLeftMargin(0.17)
c.SetRightMargin(0.025)
c.SetBottomMargin(0.14)
c.SetTopMargin(0.075)
c.SetLogy()

legend = ROOT.TLegend(0.67, 0.68, 0.9, 0.9)
legend.SetBorderSize(0)
legend.SetTextSize(basetextsize)
normalise = True

var = 'genmatched_ph_seedtime_ee'
varhists = [file.Get(var) for file in open_infiles]

for i, hist in enumerate(datahist):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetTitle('')
    newhist.SetLineColor(ROOT.kBlack)
    newhist.SetLineWidth(2)
    newhist.SetMarkerStyle(20)
    newhist.SetMarkerSize(1.05)
    newhist.SetMarkerColor(ROOT.kBlack)

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.SetMinimum(2e-4)
    newhist.SetMaximum(1.1)

    newhist.GetXaxis().SetTitle("e/#gamma seed time [ns]")
    newhist.GetXaxis().SetTitleSize(1.3*basetextsize)
    newhist.GetXaxis().SetTitleOffset(1.0)
    newhist.GetXaxis().SetLabelSize(1.3*basetextsize)
    newhist.GetXaxis().SetLabelOffset(0.005)
    newhist.GetYaxis().SetTitle("Events / 0.2 ns")
    newhist.GetYaxis().SetTitleSize(1.3*basetextsize)
    newhist.GetYaxis().SetTitleOffset(1.3)
    newhist.GetYaxis().SetLabelSize(1.3*basetextsize)
    newhist.GetYaxis().SetLabelOffset(0.005)

    newhist.Draw("e1")

    legend.AddEntry(newhist, "Z #rightarrow ee, #bf{2023}", "lep")

for i, hist in enumerate(varhists):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetLineColor(colors[i])
    newhist.SetLineWidth(3)
    newhist.SetMarkerStyle(markerstyle[i])
    newhist.SetMarkerSize(1.0)
    newhist.SetMarkerColor(colors[i])

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.SetMinimum(2e-4)
    newhist.SetMaximum(0.95)

    newhist.Draw("e1 same")

    legend.AddEntry(newhist, f'c#tau = {legendname[i]}', "lpe")

text = ROOT.TLatex()
text.SetTextSize(1.3*basetextsize)
text.SetTextFont(42)
text.SetNDC()
text.DrawLatex(0.2, 0.8525, "#bf{CMS}")
text.SetTextSize(basetextsize)
text.DrawLatex(0.2, 0.805, "#chi^{#pm} #rightarrow #chi^{0} l^{#pm} #nu")
text.SetTextSize(basetextsize)
text.DrawLatex(0.2, 0.75, "1.6 < |#eta| < 2.1")
text.SetTextSize(basetextsize)
text.SetTextAlign(31)
text.DrawLatex(0.975, 0.945, "0.41 fb^{-1} (13.6 TeV)")

legend.Draw()

line1 = ROOT.TLine(-2.4, 1.1, 3.4, 1.1)
line1.SetLineColor(ROOT.kBlack)
line1.SetLineWidth(2)
line1.Draw("SAME")

line2 = ROOT.TLine(3.4, 2e-4, 3.4, 1.1)
line2.SetLineColor(ROOT.kBlack)
line2.SetLineWidth(2)
line2.Draw("SAME")

c.Update()

plotname = f"plots/tdelay_ecal_w_d0_ee_new.pdf"
c.SaveAs(plotname)
```

```python
# L1T+HLT efficiency in seedtime

refhistvar = 'tid_elprobe_inZwindow_seedtime'
tarhistvar = 'tidhlt_elprobe_inZwindow_seedtime'
refhist = datafile.Get(refhistvar)
tarhist = datafile.Get(tarhistvar)

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")
binedges = np.array(np.arange(0.5, 2.1, 0.1), dtype=np.float64)

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 700, 700)
ROOT.gStyle.SetOptStat(0)
ROOT.gStyle.SetLineWidth(2)
c.SetLeftMargin(0.17)
c.SetRightMargin(0.025)
c.SetBottomMargin(0.155)
c.SetTopMargin(0.075)

legend = ROOT.TLegend(0.54, 0.45, 0.74, 0.5)
legend.SetBorderSize(0)
legend.SetTextSize(basetextsize)

nrefhist = refhist.Rebin(len(binedges)-1, refhistvar+'_rebinned', binedges)
ntarhist = tarhist.Rebin(len(binedges)-1, tarhistvar+'_rebinned', binedges)

dummyhist = ROOT.TH1F("dummyhist", "dummyhist", len(binedges)-1, binedges)
dummyhist.SetTitle('')
dummyhist.SetMinimum(0)
dummyhist.SetMaximum(1.0)
dummyhist.GetXaxis().SetTitle("seed time (e_{2}) [ns]")
dummyhist.GetXaxis().SetTitleSize(1.3*basetextsize)
dummyhist.GetXaxis().SetTitleOffset(1.09)
dummyhist.GetXaxis().SetLabelSize(1.3*basetextsize)
dummyhist.GetXaxis().SetLabelOffset(0.005)
dummyhist.GetYaxis().SetTitle("L1T + HLT efficiency")
dummyhist.GetYaxis().SetTitleSize(1.3*basetextsize)
dummyhist.GetYaxis().SetTitleOffset(1.275)
dummyhist.GetYaxis().SetLabelSize(1.3*basetextsize)
dummyhist.GetYaxis().SetLabelOffset(0.01)
dummyhist.Draw("e1")

eff = ROOT.TEfficiency(ntarhist, nrefhist)
eff.SetMarkerStyle(20)
eff.SetMarkerSize(1.05)
eff.SetMarkerColor(ROOT.kBlack)
eff.SetLineColor(ROOT.kBlack)
eff.SetLineWidth(1)
eff.SetTitle('')
eff.Draw("e1 same")

legend.AddEntry(eff, "e/#gamma_{2} p_{T} > 12 GeV,", "lpe")
legend.Draw()

text = ROOT.TLatex()
text.SetTextSize(1.3*basetextsize)
text.SetTextFont(42)
text.SetNDC()
text.DrawLatex(0.225, 0.835, "#bf{CMS}")
text.SetTextSize(basetextsize)
text.DrawLatex(0.59, 0.39, "seed time > 1 ns,")
text.SetTextSize(basetextsize)
text.DrawLatex(0.59, 0.33, "|#eta| < 2.1,")
text.SetTextSize(basetextsize)
text.DrawLatex(0.59, 0.27, "2023")
text.SetTextSize(basetextsize)
text.SetTextAlign(31)
text.DrawLatex(0.975, 0.945, "10.8 fb^{-1} (13.6 TeV)")

line1 = ROOT.TLine(0.5, 1.0, 2, 1.0)
line1.SetLineColor(ROOT.kBlack)
line1.SetLineWidth(2)
line1.Draw("SAME")

line2 = ROOT.TLine(2, 0, 2, 1.0)
line2.SetLineColor(ROOT.kBlack)
line2.SetLineWidth(2)
line2.Draw("SAME")

c.Update()

plotname = f"plots/hltdipho10t1ns_eff_seedtime.pdf"
c.SaveAs(plotname)
```
