```python
from datetime import datetime
import numpy as np

from IPython.display import IFrame, Image, display
import ROOT

ROOT.gROOT.SetBatch(True)
```

```python
datafilename = './data_histos.root'
datafile = ROOT.TFile(datafilename)

filenameslist = ['./hists/hist_3cm.root', './hists/hist_30cm.root', './hists/hist_3m.root']
open_infiles = [ROOT.TFile.Open(filename) for filename in filenameslist]
```

```python
# Finalise the EB plot

datahist = [datafile.Get('eleEBZ_el0_seedtime')+datafile.Get('eleEBZ_el1_seedtime')]

legendname = ['3 cm', '30 cm', '3 m']
colors = [ROOT.kRed-8, ROOT.kRed-6, ROOT.kRed]

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")
binedges = np.array(np.arange(-3, 5.2, 0.2), dtype=np.float64)

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 800, 600)
ROOT.gStyle.SetOptStat(0)
ROOT.gStyle.SetLineWidth(2)
c.SetLeftMargin(0.15)
c.SetBottomMargin(0.15)
c.SetLogy()

legend = ROOT.TLegend(0.65, 0.65, 0.88, 0.85)
legend.SetBorderSize(0)
legend.SetTextSize(0.04)
normalise = True

var = 'genmatched_ph_seedtime_eb'
varhists = [file.Get(var) for file in open_infiles]

for i, hist in enumerate(datahist):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetTitle('')
    newhist.SetLineColor(ROOT.kBlack)
    newhist.SetLineWidth(3)

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.SetMinimum(2e-4)
    newhist.SetMaximum(0.95)

    newhist.GetXaxis().SetTitle("e/#gamma seed time [ns]")
    newhist.GetXaxis().SetTicks("-")
    newhist.GetXaxis().SetTitleSize(0.045)
    newhist.GetXaxis().SetTitleOffset(-1.5)
    newhist.GetXaxis().SetLabelSize(0.045)
    newhist.GetXaxis().SetLabelOffset(-0.08)
    newhist.GetYaxis().SetTitle("normalised event count")
    newhist.GetYaxis().SetTicks("+")
    newhist.GetYaxis().SetTitleSize(0.045)
    newhist.GetYaxis().SetTitleOffset(-1.5)
    newhist.GetYaxis().SetLabelSize(0.045)
    newhist.GetYaxis().SetLabelOffset(-0.04)

    newhist.Draw("HIST SAME E1")

    legend.AddEntry(newhist, "Z #rightarrow ee, #bf{2023}", "l")

for i, hist in enumerate(varhists):
    newhist = hist.Rebin(len(binedges)-1, var+'_rebinned', binedges)

    newhist.SetLineColor(colors[i])
    newhist.SetLineWidth(3)

    if normalise:
        newhist.Scale(1.0 / newhist.Integral())
        

    newhist.SetMinimum(2e-4)
    newhist.SetMaximum(0.95)

    newhist.Draw("HIST SAME E1")

    legend.AddEntry(newhist, f'c#tau = {legendname[i]}', "l")

text = ROOT.TLatex()
text.SetTextSize(0.065)
text.SetTextFont(42)
text.SetNDC()
text.DrawLatex(0.18, 0.82, "#bf{CMS}")
text.SetTextSize(0.04)
text.DrawLatex(0.79, 0.92, "13.6 TeV")

legend.Draw()

line1 = ROOT.TLine(-3, 0.95, 5, 0.95)
line1.SetLineColor(ROOT.kBlack)
line1.SetLineWidth(2)
line1.Draw("SAME")

line2 = ROOT.TLine(5, 2e-4, 5, 0.95)
line2.SetLineColor(ROOT.kBlack)
line2.SetLineWidth(2)
line2.Draw("SAME")

c.Update()

plotname = f"plots/{var}_{now_str}.pdf"
c.SaveAs(plotname)
```
