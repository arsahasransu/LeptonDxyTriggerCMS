```python
from datetime import datetime
import numpy as np
import pickle

import ROOT

ROOT.gROOT.SetBatch(True)
```

```python
dymcfilename = './data_histos.root'
dymcfile = ROOT.TFile(dymcfilename)

bsmfilename = './hists/hist_3cm.root'
bsmfile = ROOT.TFile.Open(bsmfilename)
```

```python
def make_event_weight_file(eta, bes):

    dymch = dymcfile.Get(f'ele{eta}Z_el0_pt')+dymcfile.Get(f'ele{eta}Z_el1_pt')
    bsmh = bsmfile.Get(f'genmatched_ph_pt_{eta.lower()}')

    print(len(bes))
    print(bes)

    dymch_ntot = dymch.Integral()
    bsmh_ntot = bsmh.Integral()

    evt_wts = np.zeros(len(bes))
    i = 0
    d_w = 0.0
    b_w = 0.0
    for nbin in range(dymch.GetNbinsX()+2):

        # if nbin > 15:
        #     break

        blow = dymch.GetBinLowEdge(nbin)

        if abs(blow-bes[i]) < 1e-6:
            # print(d_w, b_w)
            if i>0:
                evt_wts[i-1] = b_w/d_w
            d_w = 0.0
            b_w = 0.0
            i += 1

        d_w += dymch.GetBinContent(nbin)/dymch_ntot
        b_w += bsmh.GetBinContent(nbin)/bsmh_ntot
        # print(nbin, d_w, b_w)

    print(len(evt_wts))
    print(evt_wts)

    for be, wt in zip(bes, evt_wts):
        print(f'{be} : {wt}')

    combined_data_dump = [bes, evt_wts]

    with open(f"{eta}_event_weights_by_pt.pkl", "wb") as f:
        pickle.dump(combined_data_dump, f)
```

```python
# For EB
bes_eb = np.concatenate((np.arange(10, 31, 1), np.arange(32, 52, 2),
                         np.arange(54, 74, 4), np.arange(80, 110, 10)))
make_event_weight_file('EB', bes_eb)
dymch = dymcfile.Get('eleEBZ_el0_pt')+dymcfile.Get('eleEBZ_el1_pt')
dymch_wt = dymcfile.Get('eleEBZ_el0_weighted_pt')+dymcfile.Get('eleEBZ_el1_weighted_pt')
bsmh = bsmfile.Get('genmatched_ph_pt_eb')

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 800, 700)
ROOT.gStyle.SetOptStat(0)
c.SetLogy()
bsmh.SetMinimum(1.0)
bsmh.SetLineColor(ROOT.kBlack)
bsmh.DrawNormalized('hist e1')
dymch.SetLineColor(ROOT.kBlue)
dymch.DrawNormalized('hist same e1')
dymch_wt.SetLineColor(ROOT.kRed)
dymch_wt.DrawNormalized('hist same e1')
```

```python
# For EE
bes_ee = np.concatenate((np.arange(10, 25, 1), np.arange(26, 32, 2),
                         np.arange(35, 50, 5), np.arange(55, 75, 10),
                         np.arange(80, 120, 20)))
make_event_weight_file('EE', bes_ee)
dymch = dymcfile.Get('eleEEZ_el0_pt')+dymcfile.Get('eleEEZ_el1_pt')
dymch_wt = dymcfile.Get('eleEEZ_el0_weighted_pt')+dymcfile.Get('eleEEZ_el1_weighted_pt')
bsmh = bsmfile.Get('genmatched_ph_pt_ee')

now_str = datetime.now().strftime("%Y_%m_%d_%H_%M_%S")

c = ROOT.TCanvas("c"+now_str, "c"+now_str, 800, 700)
ROOT.gStyle.SetOptStat(0)
c.SetLogy()
bsmh.SetMinimum(1.0)
bsmh.SetLineColor(ROOT.kBlack)
bsmh.DrawNormalized('hist e1')
dymch.SetLineColor(ROOT.kBlue)
dymch.DrawNormalized('hist same e1')
dymch_wt.SetLineColor(ROOT.kRed)
dymch_wt.DrawNormalized('hist same e1')
```
